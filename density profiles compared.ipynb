# test

```julia
# data
L = 200
α = 0.8
β = 0.8
q_defect = 0.5
defect1_length = 3
from = :center
q = lattice(L, q_defect, defect1_length, from)
tasep = Tasep(L, α, β, q)  ## example tasep with variable q
ρ0 = 0.5
J0 = 0.25
cond_init = (ρ0, J0)  ## example initial conditions;
cond_init3 = (0.5, 0.25, 0.25, 0.25)  ## example initial conditions;

# simulation

    d1,J1 = EuleroMF(tasep, cond_init)
    d2,J2 = EuleroPA(tasep, cond_init)
    d3,J3 ,_ ,_ = EuleroTRI(tasep, cond_init3)
    d,J = Gillespie_temp(tasep, cond_init;termalization_time=10^6, n_measures=10^7)
    d1 = d1[:,end] # contained the virtual sites 0 and L+1
    d2 = d2[:,end];
    
```

```julia
# save/load the data
name_data = "test_data.jld2"
@save path_data*name_data d1 d2 d3 d
#@load path_data*name_data d1 d2 d3 d

p = plot(xlabel="Sites", ylabel="ρ", title="", legend=false)
plot!(p, d1, label="Mean-Field",)
plot!(p, d2, label="Pair-Approximation", color=:darkorange)
plot!(p, d3, label="Triplet-Approximation", color=:green3)
plot!(p, d, label="Numerical solution", color=:crimson, lw=1, ls=:dot)


zoom_range = 95:105
x, y = 0.25, 0.45
width, height = 0.3, 0.4  # inset dimensions
# Inset axis
plot!(p, inset=(1, bbox(x, y, width, height)))  # inset position

# Plot the zoomed data on the inset axis (subplot 2)
plot!(p[2], zoom_range, d1[zoom_range],
    legend=false,
    titlefontsize=8,
    tickfontsize=6,
    ylims=(0.49, 0.505),
    framestyle=:box)
plot!(p[2], zoom_range, d[zoom_range], color=:crimson, lw=1, ls=:dot)
plot!(p[2], zoom_range, d2[zoom_range], color=:darkorange)
plot!(p[2], zoom_range, d3[zoom_range], color=:green3)
```
